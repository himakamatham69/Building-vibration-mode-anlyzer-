```python

# ============================================
# BUILDING VIBRATION MODE ANALYSER
# Using Eigenvalues and Eigenvectors
# ============================================

import numpy as np
import matplotlib.pyplot as plt

print("======================================")
print("   BUILDING VIBRATION MODE ANALYSER")
print("======================================")

# -------------------------------
# 1. GET INPUTS
# -------------------------------

m1 = float(input("Enter mass of Floor 1: "))
m2 = float(input("Enter mass of Floor 2: "))

k1 = float(input("Enter stiffness k1: "))
k2 = float(input("Enter stiffness k2: "))


# -------------------------------
# 2. CREATE MASS MATRIX
# -------------------------------

M = np.array([
    [m1, 0],
    [0, m2]
])


# -------------------------------
# 3. CREATE STIFFNESS MATRIX
# -------------------------------

K = np.array([
    [k1 + k2, -k2],
    [-k2, k2]
])


# -------------------------------
# 4. DISPLAY MATRICES
# -------------------------------

print("\n========== MASS MATRIX ==========")
print(M)

print("\n======= STIFFNESS MATRIX ========")
print(K)


# -------------------------------
# 5. EIGENVALUE CALCULATION
# -------------------------------

# Solve Kφ = λMφ
A = np.linalg.inv(M) @ K

eigenvalues, eigenvectors = np.linalg.eig(A)


# -------------------------------
# 6. SORT RESULTS
# -------------------------------

order = np.argsort(eigenvalues)

eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]


# -------------------------------
# 7. NATURAL FREQUENCY
# -------------------------------

omega = np.sqrt(eigenvalues)

frequency = omega / (2 * np.pi)


# -------------------------------
# 8. DISPLAY MATHEMATICAL RESULTS
# -------------------------------

print("\n======================================")
print("       EIGENVALUE RESULTS")
print("======================================")

for i in range(len(eigenvalues)):

    print(f"\nMODE {i + 1}")

    print(f"Eigenvalue (λ) = {eigenvalues[i]:.4f}")

    print(f"Angular frequency (ω) = {omega[i]:.4f} rad/s")

    print(f"Natural frequency (f) = {frequency[i]:.4f} Hz")

    print("Eigenvector:")

    print(eigenvectors[:, i])


# -------------------------------
# 9. SHOW EQUATION
# -------------------------------

print("\n======================================")
print("       GOVERNING EQUATION")
print("======================================")

print("K φ = λ M φ")

print("\nNatural frequency equation:")

print("ω = √λ")

print("\nFrequency equation:")

print("f = ω / (2π)")


# -------------------------------
# 10. VISUALIZE VIBRATION MODES
# -------------------------------

floors = np.array([0, 1, 2])

for mode in range(2):

    # Eigenvector gives the relative
    # movement of each floor
    shape = eigenvectors[:, mode]

    # Ground does not move
    displacement = np.array([
        0,
        shape[0],
        shape[1]
    ])

    plt.figure(figsize=(6, 5))

    plt.plot(
        displacement,
        floors,
        marker='o',
        linewidth=3
    )

    # Ground/reference line
    plt.axvline(
        x=0,
        linestyle='--'
    )

    plt.xlabel("Relative Displacement")

    plt.ylabel("Floor")

    plt.title(
        f"Building Vibration Mode {mode + 1}"
    )

    plt.yticks(
        [0, 1, 2],
        ["Ground", "Floor 1", "Floor 2"]
    )

    plt.grid(True)

    plt.show()


# -------------------------------
# 11. FINAL SUMMARY
# -------------------------------

print("\n======================================")
print("           PROJECT SUMMARY")
print("======================================")

print("Number of floors: 2")

print(f"Mode 1 frequency = {frequency[0]:.4f} Hz")

print(f"Mode 2 frequency = {frequency[1]:.4f} Hz")

print("\nProject completed successfully!")
```
